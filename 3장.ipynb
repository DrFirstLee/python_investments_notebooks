### <그림>그림 3-2 코스피 지수와 원/달러 환율 추이


```python


# 2. 10년 전 날짜 계산
end_date = datetime.datetime.now()
start_date = end_date - datetime.timedelta(days=365 * 10)

start_date_str = start_date.strftime('%Y-%m-%d')
end_date_str = end_date.strftime('%Y-%m-%d')

print(f"데이터 수집 기간: {start_date_str} ~ {end_date_str}")

# 3. 데이터 수집 (FinanceDataReader 사용)
kospi = fdr.DataReader('KS11', start_date_str, end_date_str)
usd_krw = fdr.DataReader('USD/KRW', start_date_str, end_date_str)

# 데이터프레임 병합 (날짜 기준으로 정렬 및 결측치 제거)
df = kospi[['Close']].rename(columns={'Close': 'KOSPI'}).join(
    usd_krw[['Close']].rename(columns={'Close': 'USD_KRW'}),
    how='inner'
)

# 4. 차트 그리기 (figsize: 12인치 x 6인치)
fig, ax1 = plt.subplots(figsize=(12, 6))

# 왼쪽 Y축: 코스피 지수 (KOSPI)
color_kospi = '#1f77b4'  # 신뢰감을 주는 네이비/블루
ax1.set_xlabel('날짜', fontsize=12, labelpad=10)
ax1.set_ylabel('코스피 지수 (KOSPI)', color=color_kospi, fontsize=12)
line1 = ax1.plot(df.index, df['KOSPI'], color=color_kospi, label='코스피 지수', alpha=0.8, linewidth=1.5)
ax1.tick_params(axis='y', labelcolor=color_kospi)
ax1.grid(True, linestyle='--', alpha=0.5)  # 배경 격자선 설정

# 오른쪽 Y축: 원/달러 환율 (USD/KRW)
ax2 = ax1.twinx()
color_usd = '#d62728'  # 시인성 높은 레드
ax2.set_ylabel('원/달러 환율 (USD/KRW)', color=color_usd, fontsize=12)
line2 = ax2.plot(df.index, df['USD_KRW'], color=color_usd, label='원/달러 환율', alpha=0.8, linewidth=1.5)
ax2.tick_params(axis='y', labelcolor=color_usd)

# 두 선의 범례(Legend)를 하나로 통합하여 좌측 상단에 표시
lines = line1 + line2
labels = [l.get_label() for l in lines]
ax1.legend(lines, labels, loc='upper left', fontsize=11, frameon=True, shadow=True)

# 차트 제목 설정 (전역 폰트가 적용되어 맑은고딕 선언 없이 잘 나옵니다)
plt.title('최근 10개년 코스피(KOSPI) 지수와 원/달러 환율 추이 비교', fontsize=15, fontweight='bold', pad=15)
fig.tight_layout()

# 차트 출력
plt.show()
```

### <그림>그림 3- 멀티팩터 포트폴리오 누적 수익률 비교(최근 6개월)

```python
import os
import sys
import warnings
from datetime import datetime, timedelta
from pathlib import Path

import matplotlib.pyplot as plt
import matplotlib.font_manager as fm
import pandas as pd
import yfinance as yf

from MultiFactor import MultiFactorUS

# 폰트 캐시 경고 무시 및 나눔바른고딕 강제 등록
warnings.filterwarnings(
    action="ignore", category=UserWarning, module="matplotlib"
)
font_path = "/usr/share/fonts/truetype/nanum/NanumBarunGothic.ttf"
fm.fontManager.addfont(font_path)

# matplotlib 전역 설정 반영
plt.rcParams["font.family"] = "NanumBarunGothic"
plt.rcParams["axes.unicode_minus"] = False

# 주요 파라미터 정의
UNIVERSE_SIZE = 500
GROUP_SIZE = 25
LOOKBACK_DAYS = 180
BENCHMARK_TICKER = "SPY"


# ==========================================
# 2. 멀티팩터 스코어 산출 및 상/하위 그룹 분리
# ==========================================
print(f"MultiFactorUS(N={UNIVERSE_SIZE})로 멀티팩터 종합점수를 산출합니다.")
mf = MultiFactorUS(N=UNIVERSE_SIZE)
data_mast = mf.get_score()

# 필수 컬럼 검증
required_columns = {"scode", "sname", "종합순위"}
missing_columns = required_columns - set(data_mast.columns)
if missing_columns:
    raise ValueError(
        f"MultiFactorUS 결과에 필요한 컬럼이 없습니다: {sorted(missing_columns)}"
    )

# 순위 정렬 후 상위/하위 그룹 분리
data_sorted = data_mast.sort_values(by="종합순위")
top_group = data_sorted.head(GROUP_SIZE).copy()
bottom_group = data_sorted.tail(GROUP_SIZE).copy()

top_tickers = top_group["scode"].tolist()
bottom_tickers = bottom_group["scode"].tolist()


# ==========================================
# 3. 데이터 수집 기간 설정 및 주가 데이터 다운로드
# ==========================================
end_date = datetime.now().strftime("%Y-%m-%d")
start_date = (datetime.now() - timedelta(days=LOOKBACK_DAYS)).strftime(
    "%Y-%m-%d"
)

all_tickers = top_tickers + bottom_tickers + [BENCHMARK_TICKER]
print(f"\n최근 6개월 주가 데이터 다운로드: {start_date} ~ {end_date}")
print(f"다운로드 대상: {len(all_tickers)}개 티커")

# yfinance 다운로드 수행
price_data = yf.download(
    all_tickers,
    start=start_date,
    end=end_date,
    auto_adjust=False,
    progress=False,
    threads=True,
)

if price_data.empty:
    raise ValueError(
        "주가 데이터를 다운로드하지 못했습니다. 네트워크 연결 또는 티커를 확인하세요."
    )

# 단일/멀티 인덱스 처리 및 결측치 전후방 보간
if isinstance(price_data.columns, pd.MultiIndex):
    close_prices = price_data["Close"].copy()
else:
    close_prices = price_data[["Close"]].rename(
        columns={"Close": all_tickers[0]}
    )

close_prices = close_prices.dropna(axis=1, how="all").ffill().bfill()
if close_prices.empty:
    raise ValueError("유효한 종가 데이터가 없습니다.")


# ==========================================
# 4. 상위/하위 그룹 동일가중 누적 수익률 계산
# ==========================================
# 상위 그룹 누적 수익률 계산
available_top = [t for t in top_tickers if t in close_prices.columns]
if not available_top:
    raise ValueError(f"상위 그룹 중 다운로드된 티커가 없습니다.")
top_daily = close_prices[available_top].pct_change().fillna(0).mean(axis=1)
top_cumulative = ((1 + top_daily).cumprod() - 1) * 100

# 하위 그룹 누적 수익률 계산
available_bottom = [t for t in bottom_tickers if t in close_prices.columns]
if not available_bottom:
    raise ValueError(f"하위 그룹 중 다운로드된 티커가 없습니다.")
bottom_daily = close_prices[available_bottom].pct_change().fillna(0).mean(axis=1)
bottom_cumulative = ((1 + bottom_daily).cumprod() - 1) * 100


# ==========================================
# 5. 시각화 (화면 출력 전용)
# ==========================================
plt.figure(figsize=(10, 6))
plt.plot(
    top_cumulative.index,
    top_cumulative.values,
    label="상위 그룹 (Top 25)",
    linewidth=2.4,
    color="#1f77b4",
)
plt.plot(
    bottom_cumulative.index,
    bottom_cumulative.values,
    label="하위 그룹 (Bottom 25)",
    linewidth=2.4,
    color="#d62728",
)

# 벤치마크(SPY) 추가
if BENCHMARK_TICKER in close_prices.columns:
    benchmark_cumulative = (
        close_prices[BENCHMARK_TICKER] / close_prices[BENCHMARK_TICKER].iloc[0]
        - 1
    ) * 100
    plt.plot(
        benchmark_cumulative.index,
        benchmark_cumulative.values,
        label="시장 전체 (SPY)",
        linewidth=1.8,
        linestyle="--",
        color="#555555",
    )

plt.title(
    "상위 그룹 vs 하위 그룹 누적 수익률 비교", fontsize=15, fontweight="bold", pad=14
)
plt.xlabel("날짜", fontsize=12)
plt.ylabel("누적 수익률 (%)", fontsize=12)
plt.grid(True, linestyle=":", alpha=0.6)
plt.legend(fontsize=11, loc="best")
plt.tight_layout()

# 화면에 차트 출력 후 컨텍스트 해제
plt.show()
```
