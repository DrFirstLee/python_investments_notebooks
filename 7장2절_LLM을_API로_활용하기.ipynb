```python
from google import genai
from google.genai import types
# 1. API 키 설정
client = genai.Client(api_key="{나의_API_KEY}")

#. 2. Search 설정 : 그라운딩 : 검색기능을 연결해줍니다
grounding_tool = types.Tool(
    google_search=types.GoogleSearch()
)

config = types.GenerateContentConfig(
    tools=[grounding_tool]
)

# Prompt 세팅
prompt = """
# 중요 지시 (필수 제약 조건)
- 당신은 내부 지식만으로는 현재 주가와 최신 기업 실적을 정확히 알지 못합니다.
- 반드시 제공된 Google 검색 기능(Search Tool)을 실행하여 2026년 6월 현재의 삼성전자와 SK하이닉스 관련 최신 뉴스, 공시, 주가 데이터를 실시간으로 조회한 뒤, 그 결과를 바탕으로 보고서를 작성해야 합니다. 검색을 생략하지 마세요.

# 지시문
- 당신은 증권사에서 반도체 업종을 담당하고 있는 리서치 연구원입니다.
- 삼성전자와 SK하이닉스 주식 투자자를 대상으로, 두 주식의 투자 전망을 비교하는 보고서를 작성해주세요.

# 제약 조건
- 삼성전자와 SK하이닉스와 관련된 뉴스, 리서치보고서, 재무데이터를 수집해서 분석합니다.
- 반드시 최신 정보를 찾아야 합니다.
- 오늘은 [2026.06.25] 입니다. Google 검색으로 자료를 수집합니다.
- 보고서는 객관적인 데이터와 발표자료를 근거로 출력해주세요.
- 보고서는 삼성전자와 SK하이닉스의 향후 주가 전망을 비교하는 이슈에 집중합니다.

# 출력형식
- 최근의 삼성전자와 SK하이닉스의 주가 추이와 실적을 요약합니다.
- 삼성전자와 SK하이닉스의 주요 이슈를 정리하고, 향후 실적과 주가에 미칠 영향에 대한 분석 결과를 나열합니다.
- 현재 삼성전자와 SK하이닉스의 주가를 나열하고, 목표주가를 제안합니다.
- 최종적으로 투자 전망 측면에서 1개 기업을 선정하고, 선정 사유와 유의사항을 출력합니다.
"""

# 3. API로 묻기!
response = client.models.generate_content(
    model="gemini-2.5-flash",
    contents=prompt,
    config=config,
)


# 4. 결과에 참고된 검색 데이터 소스보기
def add_citations(response):
    text = response.text
    supports = response.candidates[0].grounding_metadata.grounding_supports
    chunks = response.candidates[0].grounding_metadata.grounding_chunks

    # Sort supports by end_index in descending order to avoid shifting issues when inserting.
    sorted_supports = sorted(supports, key=lambda s: s.segment.end_index, reverse=True)

    for support in sorted_supports:
        end_index = support.segment.end_index
        if support.grounding_chunk_indices:
            # Create citation string like [1](link1)[2](link2)
            citation_links = []
            for i in support.grounding_chunk_indices:
                if i < len(chunks):
                    uri = chunks[i].web.uri
                    citation_links.append(f"[{i + 1}]({uri})")

            citation_string = ", ".join(citation_links)
            text = text[:end_index] + citation_string + text[end_index:]

    return text

text_with_citations = add_citations(response)
print(text_with_citations)
```
