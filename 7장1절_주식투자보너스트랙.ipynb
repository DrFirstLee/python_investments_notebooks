```python
from google import genai

# 1. 발급받은 API 키를 설정합니다.
#client = genai.Client(api_key= "여기에_발급받은_API_키를_넣으세요")
client = genai.Client(api_key= "{key입력}")

# 2. 모델 설정 및 질문 (client.models.generate_content 사용)
response = client.models.generate_content(
    model='gemini-2.5-flash',
    contents="주식 투자의 기본 원칙 3가지만 알려줘"
)

# 4. 답변을 출력합니다.
print(response.text)
```

```python
from google import genai
from google.genai import types

# 1. API 키 설정
client = genai.Client(api_key="{key입력}")

#. 2. Search 설정 : 그라운딩 : 검색기능을 연결해줍니다
grounding_tool = types.Tool(
    google_search=types.GoogleSearch()
)

config = types.GenerateContentConfig(
    tools=[grounding_tool]
)

# Prompt 세팅
prompt = """
# 지시문
- 당신은 증권사에서 반도체 업종을 담당하고 있는 리서치 연구원입니다.
- 삼성전자와 SK하이닉스 주식 투자자를 대상으로, 두 주식의 투자전망을 비교하는 보고서를 작성해주세요.

# 제약 조건
- 삼성전자와 SK하이닉스와 관련된 뉴스, 리서치보고서, 재무데이터를 수집해서 분석합니다.
- 반드시 최신 정보를 찾아야 합니다.
- 오늘은 [2026.06.25] 입니다. Google 검색으로 자료를 수집합니다.
- 보고서는 객관적인 데이터와 발표자료를 근거로 출력해주세요.
- 보고서는 삼성전자와 SK하이닉스의 향후 주가 전망을 비교하는 이슈에 집중합니다.

# 출력형식
- 최근의 삼성전자와 SK하이닉스의 주가 추이와 실적을 요약합니다.
- 삼성전자와 SK하이닉스의 주요 이슈를 정리하고, 향후 실적과 주가에 미칠 영향에 대한 분석 결과를 나열합니다.
- 현재 삼성전자와 SK하이닉스의 주가를 나열하고, 목표주가를 제안합니다.
- 최종적으로 투자전망 측면에서 1개 기업을 선정하고, 선정 사유와 유의사항을 출력합니다.
"""

# 3. API로 묻기!
response = client.models.generate_content(
    model="gemini-2.5-flash",
    contents=prompt,
    config=config,
)

# 4. 결과 출력
print(response.text)
```
